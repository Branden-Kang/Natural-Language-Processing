# 2. Embeddings — How AI Understands Meaning

```python
import torch
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Provide documents in text as input
corpus = [
    "Put your input text",
]

corpus_embeddings = embedder.encode_document(corpus, convert_to_tensor=True)

# Query sentences:
queries = [
    "Put your queries here",
]

# Get 2 sentences of the input document for each query sentence based on cosine similarity
top_k = min(2, len(corpus))
for query in queries:
    query_embedding = embedder.encode_query(query, convert_to_tensor=True)

    # Used cosine-similarity and torch.topk to find the highest 2 scores
    similarity_scores = embedder.similarity(query_embedding, corpus_embeddings)[0]
    scores, indices = torch.topk(similarity_scores, k=top_k)

    print("\nQuery:", query)
    print("Top 2 most similar sentences in corpus:")

    for score, idx in zip(scores, indices):
        print(f"(Score: {score:.4f})", corpus[idx])
```

# 7. Temperature — Controls Creativity vs Accuracy
Low temperature:
- More accurate
- More predictable
- Good for code and facts

High temperature:
- More creative
- More diverse
- Good for writing and ideas

Typical usage:
- 0.2 → Coding
- 0.4 → Technical answers
- 0.7 → Normal writing
- 1.0 → Creative writing

# 9. RAG — Retrieval Augmented Generation
Flow:
1. User asks question
2. Convert question into embedding
3. Search vector database
4. Retrieve relevant documents
5. Send documents to model
6. Model generates answer

This reduces hallucination and allows AI to use your own data.
Most real AI applications use RAG:
- Chat with PDF
- Company knowledge bots
- Documentation assistants
- Support bots
- Internal search tools
