# 1. OpenAI SDK (Built-in MCP Support)


````python
from openai import OpenAI

client = OpenAI()
response = client.responses.create(
    model="gpt-4.1",
    input="Check server health and restart if CPU > 85%",
    tools=[
        {
            "type": "mcp",
            "server_label": "devops_server",
            "server_url": "https://mcp.company.com"
        }
    ]
)
print(response.output_text)
````

# 2. Composio


````python
from composio_openai import ComposioToolSet
from openai import OpenAI

client = OpenAI()
toolset = ComposioToolSet(api_key="YOUR_KEY")
tools = toolset.get_tools()
response = client.responses.create(
    model="gpt-4.1",
    input="Create a GitHub issue for the payment API bug",
    tools=tools
)
print(response.output_text)
````

# 3. Google ADK


````python
from google.adk import Agent

agent = Agent(
    model="gemini-pro",
    mcp_servers=["https://mcp.enterprise.com"]
)
response = agent.run("Fetch Q4 revenue breakdown by region")
print(response)
````

# 4. MCP Python SDK


````python
from mcp.server import MCPServer

server = MCPServer()
@server.tool()
def calculate_tax(amount: float) -> float:
    return amount * 0.18
server.run()
````

# 5. MCP TypeScript SDK


````python
import { MCPServer } from "@mcp/sdk";

const server = new MCPServer();
server.tool("greet", async ({ name }) => {
  return `Hello ${name}`;
});
server.listen(3000);
````
