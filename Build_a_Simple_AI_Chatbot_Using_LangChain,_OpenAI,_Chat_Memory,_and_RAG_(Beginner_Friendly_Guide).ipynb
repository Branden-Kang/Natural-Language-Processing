# Step 2: Create a Basic Chatbot


```python
from langchain.chat_models import ChatOpenAI
from langchain.schema import HumanMessage
chat = ChatOpenAI(model="gpt-3.5-turbo")
response = chat([
    HumanMessage(content="Hello, who are you?")
])
print(response.content)
```

# Step 3: Add Chat Memory (Context Awareness)


```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain
llm = ChatOpenAI(model="gpt-3.5-turbo")
memory = ConversationBufferMemory()
conversation = ConversationChain(
    llm=llm,
    memory=memory
)
print(conversation.predict(input="Hi, my name is Alex"))
print(conversation.predict(input="What is my name?"))
```

# Step 4: Understanding RAG (Retrieval Augmented Generation)
RAG works in three steps:
1. Store your documents as embeddings
2. Retrieve relevant chunks based on the question
3. Pass them to the AI model as context


# Step 5: Load Your Documents


```python
from langchain.document_loaders import TextLoader
loader = TextLoader("data.txt")
documents = loader.load()
```

# Step 6: Create Embeddings and Vector Store

```python
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import FAISS
embeddings = OpenAIEmbeddings()
vectorstore = FAISS.from_documents(
    documents,
    embedding=embeddings
)
```

# Step 7: Build a Retrieval Chain


```python
from langchain.chains import ConversationalRetrievalChain
qa_chain = ConversationalRetrievalChain.from_llm(
    llm=ChatOpenAI(model="gpt-3.5-turbo"),
    retriever=vectorstore.as_retriever(),
    memory=ConversationBufferMemory(
        memory_key="chat_history",
        return_messages=True
    )
)
```

# Step 8: Ask Questions From Your Data


```python
result = qa_chain({
    "question": "What is this document about?"
})
print(result["answer"])
```
