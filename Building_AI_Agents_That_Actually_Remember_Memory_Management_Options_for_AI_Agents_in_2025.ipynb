# LangChain Example:

```python
from langchain import OpenAI
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory

llm = OpenAI(model_name="gpt-4o")
memory = ConversationBufferMemory()
conversation = ConversationChain(llm=llm, memory=memory)
conversation.run("My name is Pankaj.")
conversation.run("Do you remember my name?")
```

# Pinecone Vector Example:

```python
import pinecone
from langchain.embeddings import OpenAIEmbeddings

pinecone.init(api_key="YOUR_API_KEY")
index = pinecone.Index("agent-memory")
emb = OpenAIEmbeddings()
# Save info
vec = emb.embed_query("Ravi loves cricket.")
index.upsert([("ravi1", vec, {"text":"Ravi loves cricket."})])
# Retrieve info
qvec = emb.embed_query("What sport does Ravi love?")
result = index.query(qvec, top_k=1)
# result["matches"][0]["metadata"]["text"] => "Ravi loves cricket."
```

# Agno Memory Example

```python
from agno.agent import Agent
from agno.models.openai import OpenAIChat

agent = Agent(
    model=OpenAIChat(id="gpt-4o"),
    memory="session",    # Keeps conversation context
    storage="persistent" # For long-term recall
)
agent.print_response("My hometown is Bengaluru.")
agent.print_response("Where is my hometown?")
# Agent answers: "Bengaluru"
```
