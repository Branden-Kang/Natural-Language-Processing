# My First AI Assistant: Automating Emails

```python
import openai
import argparse
openai.api_key = "your_api_key"
parser = argparse.ArgumentParser(description="AI Email Responder")
parser.add_argument("--context", help="Previous email content")
args = parser.parse_args()
prompt = f"""
You are my assistant. Read this email and write a professional response:
Email:
{args.context}
"""
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ],
    temperature=0.5
)
print(response.choices[0].message.content)
```

# Summarizing YouTube Lectures With AI

```python
import re
from youtube_transcript_api import YouTubeTranscriptApi
import openai
def extract_video_id(url):
    match = re.search(r"(?:v=|\/)([0-9A-Za-z_-]{11}).*", url)
    return match.group(1) if match else None
def get_summary(video_url):
    video_id = extract_video_id(video_url)
    transcript = YouTubeTranscriptApi.get_transcript(video_id)
    text = " ".join([t["text"] for t in transcript])
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": f"Summarize this: {text}"}]
    )
    return response.choices[0].message.content
print(get_summary("https://www.youtube.com/watch?v=dQw4w9WgXcQ"))
```

# Organizing My PDFs with AI Embeddings

```python
from sentence_transformers import SentenceTransformer
import fitz
import os
model = SentenceTransformer("all-MiniLM-L6-v2")
def extract_text(pdf_path):
    doc = fitz.open(pdf_path)
    return " ".join([page.get_text() for page in doc])
folder = "papers"
texts = [extract_text(os.path.join(folder, f)) for f in os.listdir(folder)]
embeddings = model.encode(texts)
print("Generated embeddings for", len(embeddings), "papers.")
```

# Building a Multimodal Search Tool

```python
from transformers import CLIPProcessor, CLIPModel
from PIL import Image
import torch
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
image = Image.open("chart.png")
inputs = processor(text=["GDP growth chart"], images=image, return_tensors="pt", padding=True)
outputs = model(**inputs)
logits_per_image = outputs.logits_per_image
print("Similarity score:", logits_per_image)
```

# My First RAG (Retrieval-Augmented Generation) System

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
docs = ["AI is transforming healthcare.", "Python is great for automation."]
embeddings = model.encode(docs)
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(np.array(embeddings))
query = "What is Python used for?"
q_embed = model.encode([query])
D, I = index.search(np.array(q_embed), k=1)
print("Answer:", docs[I[0][0]])
```
