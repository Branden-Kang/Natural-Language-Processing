# Step 2: Build a Simple Agentic AI Agent


````python
from crewai import Agent, Task, Crew

import os
openai_api_key = "not_an_intern"
os.environ["OPENAI_API_KEY"] = openai_api_key
os.environ["OPENAI_MODEL_NAME"] = 'gpt-5.5'

# -----------------------------
# AGENT 1: Researcher
# -----------------------------
# Purpose:
# This agent is responsible for researching and summarizing
# the latest trends and developments in AI.
#
# Think of this agent as a research analyst
# who gathers important information.
researcher = Agent(
    role="Researcher",
    goal="Summarize the latest trends in AI.",
    backstory="You are an AI assistant tasked with distilling complex topics into clear insights.",
    verbose=True
)

# -----------------------------
# AGENT 2: Writer
# -----------------------------
# Purpose:
# This agent takes the research findings
# and converts them into a readable report.
#
# Think of this agent as a technical writer
# who transforms raw information into
# structured and engaging content.
writer = Agent(
    role="Writer",
    goal="Generate a concise weekly report based on researcher's findings.",
    backstory="You are a skilled writer who transforms data into actionable insights.",
    verbose=True
)


# -----------------------------
# TASK 1: Research Task
# -----------------------------
# Purpose:
# Assigns the Researcher agent the job of
# analyzing recent AI developments.
#
# Expected Output:
# A bullet-point summary of important AI trends.
research_task = Task(
    description=("Analyze recent AI developments and summarize key points."),
    expected_output="A bullet-point summary of the latest AI trends.",
    agent=researcher
)

# -----------------------------
# TASK 2: Writing Task
# -----------------------------
# Purpose:
# Assigns the Writer agent the responsibility
# of creating a complete report using the
# research generated from Task 1.
#
# Context:
# The writer receives the output generated
# by the research task as input context.
#
# Expected Output:
# A polished 500-word weekly AI report.
writing_task = Task(
    description="Write a 500-word report summarizing the research.",
    expected_output="A bullet-point summary of the latest AI trends.",
    context = [research_task],
    agent=writer
)

# Create the crew and execute tasks
crew = Crew(agents=[researcher, writer], tasks=[research_task, writing_task])
result = crew.kickoff()

from IPython.display import Markdown
Markdown(result.raw)
````
