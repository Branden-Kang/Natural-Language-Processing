```python
from openai import OpenAI
client = OpenAI()

tools = [{
  "name": "create_ticket",
  "input_schema": {
    "type": "text",  # plaintext instead of JSON
    "pattern": r"^SEV[1-3]\s+[A-Z_]{2,20}\s+.+$"
  },
  "description": "Format: <SEV1|SEV2|SEV3> <SYSTEM> <TITLE>"
}]
resp = client.responses.create(
  model="gpt-5",
  reasoning={"effort": "medium"},
  verbosity="medium",
  tools=tools,
  input=[{"role": "user", "content": "Raise incident: SEV2 PAYMENTS Timeout on checkout at peak."}]
)
print(resp.output_text)
```
