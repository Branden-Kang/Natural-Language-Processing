# Step 2: Build the Service Logic

````python
pip install youtube_transcript_api
````

````python
# src/service.py
from youtube_transcript_api import YouTubeTranscriptApi

class YouTubeTranscriptService:
    def get_transcript(self, video_id: str):
        return YouTubeTranscriptApi.get_transcript(video_id)
````

````python
# src/utils/video.py
import re

def extract_video_id(url: str) -> str:
    match = re.search(r"v=([\w-]+)", url)
    return match.group(1) if match else url
````

# Step 3: Register Your Tool via MCP

````python
# server.py
#!/usr/bin/env uv
# /// script
# dependencies = ["mcp-cli", "youtube-transcript-api"]
# ///

from mcp import MCPServer
from src.service import YouTubeTranscriptService
from src.utils.video import extract_video_id
service = YouTubeTranscriptService()
server = MCPServer()
@server.tool()
def get_transcript(video_url: str):
    video_id = extract_video_id(video_url)
    return service.get_transcript(video_id)
if __name__ == "__main__":
    server.run()
````

# Step 5: Connect Your MCP Server to Claude Desktop

1. Open Claude Desktop → Settings → Developer → Edit Config
2. Add your server config:
```
{
  "mcpServers": {
    "youtube": {
      "command": "uv",
      "args": ["run", "server.py"],
      "cwd": "/absolute/path/to/mcp_server"
    }
  }
}
```
3. Restart Claude Desktop.
