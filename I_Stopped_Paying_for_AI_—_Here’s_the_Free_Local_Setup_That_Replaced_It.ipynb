# Bonus: Use It From Code (Because You Were Going To Ask)
```
curl http://localhost:11434/api/generate -d '{
  "model": "gemma4:e4b",
  "prompt": "Summarize why flaky tests are the worst in one sentence",
  "stream": false
}'
```

````python
import requests

OLLAMA_URL = "http://localhost:11434/api/generate"
def triage_test_failure(log_path: str) -> str:
    """Send a failing test log to local Gemma 4 for a first-pass triage.
    Everything stays on-device - safe for logs with internal data."""
    with open(log_path, "r") as f:
        log_content = f.read()[-8000:]  # last 8K chars is usually where the pain is
    prompt = f"""You are a senior QA engineer. Analyze this failed test log.
Classify the failure as one of: PRODUCT_BUG, TEST_BUG, ENV_ISSUE, FLAKY.
Give a one-line reason and one suggested next step.
Log:
{log_content}"""
    response = requests.post(OLLAMA_URL, json={
        "model": "gemma4:e4b",
        "prompt": prompt,
        "stream": False,
        # Google's recommended sampling params for Gemma 4
        "options": {"temperature": 1.0, "top_p": 0.95, "top_k": 64}
    })
    return response.json()["response"]
if __name__ == "__main__":
    print(triage_test_failure("./reports/checkout_flow_failure.log"))
````
