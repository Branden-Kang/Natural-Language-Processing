# RunnableSequence (Most Important)

````python
from langchain_core.runnables import RunnableSequence

chain = RunnableSequence(prompt, model, parser)

result = chain.invoke({"topic": "AI"})

print(result)
````

# RunnableSequence using Pipe (LCEL)
- LCEL (LangChain Expression Language)

````python
chain = prompt | model | parser
````

````python
from langchain_core.runnables import RunnableParallel

parallel_chain = RunnableParallel(
    tweet = prompt1 | model | parser,
    linkedin = prompt2 | model | parser
)

result = parallel_chain.invoke({"topic": "AI"})

print(result["tweet"])
print(result["linkedin"])
````

# RunnableLambda
```
LLM → text → lambda → word count
```

````python
from langchain_core.runnables import RunnableLambda

def count_words(text):
    return len(text.split())

word_counter = RunnableLambda(count_words)
````

```
chain = prompt | model | parser | word_counter
```

# chain = prompt | model | parser | word_counter
```
Prompt → LLM → Parser → Branch
                    → summarize
                    → passthrough
```

````python
from langchain_core.runnables import RunnableBranch

branch = RunnableBranch(
    (
        lambda x: len(x.split()) > 500,
        summary_prompt | model | parser
    ),
    RunnablePassthrough()

chain = prompt | model | parser | branch
````
