# What Are Tools?


````python
from langchain.tools import tool

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
````

## Step 1: Tool Binding


````python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()

llm_with_tools = llm.bind_tools([multiply])
````

## Step 2: Tool Calling
```
Tool Calling is when the LLM decides that a tool should be used to solve a problem.
```


```
{
  "tool_name": "multiply",
  "arguments": {
    "a": 3,
    "b": 10
  }
}
```

## Step 3: Tool Execution


````python
result = multiply.invoke({"a": 3, "b": 10})
# 30
````

# Tool Message

LangChain supports different message types:
- HumanMessage
- AIMessage
- SystemMessage
- ToolMessage


# Full Tool Workflow
```
User Query
     ↓
LLM decides a tool is needed
     ↓
LLM generates Tool Call
     ↓
Developer executes the tool
     ↓
Tool returns result
     ↓
Result is sent back to LLM
     ↓
LLM generates final answer
```
