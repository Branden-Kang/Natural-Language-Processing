```python
import ollama

response = ollama.systemone(
    model="nimble",
    state={
        "ticket": (
            "Checkout has returned 500 errors since 09:00. "
            "Several customers are blocked."
        )
    },
    questions={
        "team": {
            "type": "choice",
            "instructions": "Which team should own this incident?",
            "criteria": {
                "billing": "Payments, invoices, or refunds",
                "platform": "Application or infrastructure failure",
                "support": "Non-technical customer support issue",
            },
        },
        "production_incident": {
            "type": "noul",
            "instructions": "Is this an active production incident?",
        },
        "urgency": {
            "type": "score",
            "instructions": "How urgent is the incident?",
            "criteria": ["Routine", "Soon", "Urgent"],
        },
    },
)
print(response.answers["team"])
print(response.answers["production_incident"])
print(response.answers["urgency"])
```
