```python
from llama_cpp import Llama

deepSeekModel = Llama.from_pretrained(repo_id="unsloth/DeepSeek-R1-Distill-Llama-8B-GGUF",
                                filename="DeepSeek-R1-Distill-Llama-8B-Q4_K_M.gguf",
                                verbose=False
                               )
```

```python
deepSeekPrompt = """
You are an AI assistant, who is ought to answer the query of the user by carefully performing reasoning on the query. Keep the answer short and concise.

### Instruction:
{prompt}

### Response:
"""
```

```python
resp = deepSeekModel(deepSeekPrompt.format(prompt="What is Yamanaka Factor in context of LLMs?"),
           max_tokens=256,
           stop=["Q:", "\n"]
       )
resp
```
