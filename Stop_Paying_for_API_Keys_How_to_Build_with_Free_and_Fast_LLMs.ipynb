# Step-by-Step: Getting Your Free API Key


Step 1:Go to console.groq.com and sign in

Step 2: On the right side, click on API Keys.

Step 3: Click the black Create API Key button. Give it a meaningful name

Step 4: Copy the key immediately.

# Choosing the Right Model
- Llama 3.1 8B Instant
- Llama 3.3 70B Versatile
- Qwen 2.5 Coder (32B)
- Whisper Large V3

```python
import os
from groq import Groq

# 1. Initialize the client
# (Make sure GROQ_API_KEY is set in your environment variables)
client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

# 2. Generate response
response = client.chat.completions.create(
    messages=[
        {"role": "system", "content": "You are a precise technical assistant."},
        {"role": "user", "content": "Explain the difference between a GPU and an LPU in two sentences."}
    ],
    # The Llama 3.1 8B model is the workhorse of the free tier
    model="llama-3.1-8b-instant",
)
print(response.choices[0].message.content)
```
