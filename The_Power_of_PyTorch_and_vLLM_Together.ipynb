# Using PyTorch and vLLM

```python
from vllm import LLM, SamplingParams


llm = LLM(
    model="google/gemma-3-1b-it",         # Any Hugging Face supported model
    trust_remote_code=True,               # Needed for custom model architectures
    dtype="auto",                         # Automatically choose best dtype
)
sampling_params = SamplingParams(temperature=0.7, top_p=0.9)
prompts = ["Tell me a joke about AI."]
results = llm.generate(prompts, sampling_params)
# Accessing output properly (important!)
for result in results:
    print(result.outputs[0].text)
```
