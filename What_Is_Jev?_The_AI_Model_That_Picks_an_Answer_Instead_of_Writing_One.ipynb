```python
from typesafe_sdk import Choice, Noul, TypeSafeClient

state = "Task: clean up the build output\nCommand: rm -rf ~/projects/app/build"

with TypeSafeClient(model="jev-1.13.0") as client:
    response = client.system_one(
        state=state,
        questions={
            "deletes_files": Noul(
                instructions="The command deletes or overwrites files",
            ),
            "touches": Choice(
                instructions="What the command touches",
                criteria={
                    "project": "Only files inside the project folder",
                    "home": "Files in the home folder outside the project",
                    "system": "System files or settings",
                    "nothing": "Nothing on disk",
                },
            ),
        },
    )

deletes = response.nouls["deletes_files"].noul
where = response.choices["touches"].choice

if deletes > 0.9 or where == "system":
    ask_before_running()
```
