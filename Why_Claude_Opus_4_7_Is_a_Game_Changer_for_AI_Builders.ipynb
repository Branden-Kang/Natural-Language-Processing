# Model lineup context

- Claude Haiku: fast, cheap, good for routing and lightweight tasks.
- Claude Sonnet: balanced default, good for most app workloads.
- Claude Opus: maximum reasoning quality, best for hard tasks.

# Using Opus 4.7 For Coding

```python
import anthropic
client = anthropic.Anthropic(api_key="YOUR_API_KEY")
prompt = "Write a small Python function that validates an email address using a regex. Return only the code."
message = client.messages.create(
  model="claude-3-opus-4.7",
  max_tokens=300,
  temperature=0.2,
  messages=[{"role": "user", "content": prompt}]
)
print(message.content[0].text)
```

# Tool Use And Agents With Opus 4.7

```python
tools = [
  {
    "name": "get_user_balance",
    "description": "Fetch the current balance for a user by ID.",
    "input_schema": {
      "type": "object",
      "properties": {"user_id": {"type": "string"}},
      "required": ["user_id"]
    }
  }
]
message = client.messages.create(
  model="claude-3-opus-4.7",
  max_tokens=300,
  tools=tools,
  messages=[{"role": "user", "content": "What is the balance for user 42?"}]
)
print(message)
```

# Opus 4.7 In Retrieval And RAG Systems

```python
system_prompt = "You are a careful assistant. Only use the provided context. If the answer is not in the context, say you do not know."
user_prompt = "Question: How does our refund policy handle digital products?"
context = "<DOC_1>...</DOC_1>\n<DOC_2>...</DOC_2>"
message = client.messages.create(
  model="claude-3-opus-4.7",
  max_tokens=500,
  temperature=0,
  messages=[
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": f"Context:\n{context}\n\n{user_prompt}"}
  ]
)
print(message.content[0].text)
```

# Prompting Patterns That Work Well With 4.7

```python
system_prompt = "You are a senior backend engineer. Think carefully, but respond with only the final code or answer."
user_prompt = "Design a simple REST API schema for a todo app. Return only JSON with endpoints and methods."
message = client.messages.create(
  model="claude-3-opus-4.7",
  max_tokens=400,
  temperature=0.3,
  messages=[
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
  ]
)
print(message.content[0].text)
```
